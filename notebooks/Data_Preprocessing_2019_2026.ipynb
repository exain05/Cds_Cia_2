4. Data Preprocessing and Cleaning

Purpose:
Prepare the 2019–2026 WFP food-price datasets for reliable analysis by standardising variables, handling data-quality issues, applying appropriate filters, and creating an analysis-ready master dataset.

4.1 Date Standardisation

```python
# Convert the date column into proper datetime format
for year, df in dfs.items():
    df["date"] = pd.to_datetime(df["date"])
```

```python
# Check that the date column is now in datetime format
for year, df in dfs.items():
    print(year, ":", df["date"].dtype)
```

```python
# Check the earliest and latest date for each year
for year, df in dfs.items():
    print(year, ":", df["date"].min(), "to", df["date"].max())
```

4.2 Creating a year Variable

```python
# Extract the year from the date column
for year, df in dfs.items():
    df["year"] = df["date"].dt.year
```

```python
# Check the years stored in each dataset
for year, df in dfs.items():
    print(year, ":", df["year"].unique())
```

```python
# Display the first five dates and their corresponding year
dfs[2019][["date", "year"]].head()
```

4.3  Examining Price Types

```python
# Check price types across all years
for year, df in dfs.items():
    print("\n", year)
    print(df["pricetype"].value_counts())
```

4.4 Examining Price Flags

```python
# Check price flags across all years
for year, df in dfs.items():
    print("\n", year)
    print(df["priceflag"].value_counts())
```

4.5 Examining Food Categories

```python
# Check food categories across all years
for year, df in dfs.items():
    print("\n", year)
    print(df["category"].value_counts())
```

```python
# Remove non-food observations from all yearly datasets
for year, df in dfs.items():
    dfs[year] = df[df["category"] != "non-food"].copy()
```

```python
# Check that non-food observations have been removed
for year, df in dfs.items():
    print(year, ":", df["category"].unique())
```

4.6 Measurement Units

```python
# Check the most common measurement units across all years
for year, df in dfs.items():
    print("\n", year)
    print(df["unit"].value_counts().head(15))
```

```python
# Count the number of different measurement units in each year
for year, df in dfs.items():
    print(year, ":", df["unit"].nunique(), "different units")
```

4.7 Handling Missing USD Prices

Missing USD prices were retained in the master dataset because the original local-currency price values are available. No estimated exchange rates were introduced. For analyses requiring USD-based comparisons, observations with missing usdprice will be excluded.

4.8  Basic Price Validation

```python
# Check for zero or negative prices
for year, df in dfs.items():
    print(year, ":", (df["price"] <= 0).sum())
```

```python
# Check for zero or negative USD prices
for year, df in dfs.items():
    print(year, ":", (df["usdprice"].fillna(0) <= 0).sum())
```

```python
# Check actual non-missing USD prices that are zero or negative
for year, df in dfs.items():
    invalid_usd = df["usdprice"].notna() & (df["usdprice"] <= 0)
    print(year, ":", invalid_usd.sum())
```

```python
# Check actual zero and negative local prices
for year, df in dfs.items():
    invalid_price = df["price"] <= 0
    print("\n", year)
    print(df.loc[invalid_price, ["commodity", "unit", "price", "currency"]].head())
```

RESULT: All observations across the 2019–2026 datasets have positive local-currency prices. No zero or negative price values were found, so no records were removed for invalid local prices.

4.9 Combining the 8 datasets

```python
# Combine all cleaned yearly datasets
master_df = pd.concat(dfs.values(), ignore_index=True)

print("Master dataset shape:", master_df.shape)
```

```python
# Check that all eight years are present
print(master_df["year"].value_counts().sort_index())
```

4.10  Final Validation

```python
print("Shape:", master_df.shape)
print("Duplicate rows:", master_df.duplicated().sum())
print("Missing values:")
print(master_df.isnull().sum())
print("Years:", sorted(master_df["year"].unique()))
```

Result: The final master dataset contains 2,945,227 observations covering 2019–2026. No duplicate rows were found. The main price, date, country, commodity and category fields contain no missing values. Missing USD prices and geographic fields were retained because they do not prevent the core food-price analysis. All eight years are present in the final dataset.
